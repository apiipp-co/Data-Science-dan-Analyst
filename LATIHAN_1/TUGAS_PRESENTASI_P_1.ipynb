# Data Cleaning

```python
# Salin dataset
clean_df = df.copy()
```

```python
# HANDLE MISSING VALUE

# Isi children kosong dengan 0
clean_df['children'] = clean_df['children'].fillna(0)

# Isi country kosong dengan negara terbanyak
clean_df['country'] = clean_df['country'].fillna(
    clean_df['country'].mode()[0]
)

# Isi agent kosong dengan 0
clean_df['agent'] = clean_df['agent'].fillna(0)

# Isi company kosong dengan 0
clean_df['company'] = clean_df['company'].fillna(0)
```

```python
# HAPUS DATA DUPLIKAT
clean_df.drop_duplicates(inplace=True)
```

```python
# HAPUS NILAI TIDAK VALID
clean_df = clean_df[clean_df['adr'] >= 0]

q99 = clean_df['adr'].quantile(0.99)

clean_df = clean_df[
    clean_df['adr'] <= q99
]

clean_df = clean_df[
    clean_df['adults'] >= 0
]

clean_df = clean_df[
    clean_df['children'] >= 0
]

clean_df = clean_df[
    clean_df['babies'] >= 0
]
```

# Hapus Data Duplicat

```python
# Mengubah tipe data agar konsisten.

clean_df['children'] = clean_df['children'].astype('int64')

clean_df['agent'] = clean_df['agent'].astype('int64')

clean_df['company'] = clean_df['company'].astype('int64')
```

# Hapus Outliner ADR

```python
# Menghapus nilai negatif dan data tidak logis.

clean_df = clean_df[
    clean_df['adr'] >= 0
]

clean_df = clean_df[
    clean_df['adults'] >= 0
]

clean_df = clean_df[
    clean_df['children'] >= 0
]

clean_df = clean_df[
    clean_df['babies'] >= 0
]
```

# Ubah Tipe Data Tanggal

```python
# Mengubah format tanggal agar mudah dianalisis.

clean_df['reservation_status_date'] = pd.to_datetime(
    clean_df['reservation_status_date']
)
```

# Rapikan Teks

```python
# Menghapus spasi berlebih pada kategori.

clean_df['hotel'] = clean_df['hotel'].str.strip()

clean_df['meal'] = clean_df['meal'].str.strip()

clean_df['market_segment'] = clean_df[
    'market_segment'
].str.strip()

clean_df['distribution_channel'] = clean_df[
    'distribution_channel'
].str.strip()
```

# Feature Enginer

```python
# Total malam menginap
clean_df['total_nights'] = (
    clean_df['stays_in_weekend_nights'] +
    clean_df['stays_in_week_nights']
)

# Total tamu
clean_df['total_guest'] = (
    clean_df['adults'] +
    clean_df['children'] +
    clean_df['babies']
)

# Total histori booking pelanggan
clean_df['customer_history'] = (
    clean_df['previous_bookings_not_canceled'] +
    clean_df['previous_cancellations']
)
```

# Hapus Booking Tanpa Tamu

```python
# Menghapus booking tanpa tamu.

clean_df = clean_df[
    clean_df['total_guest'] > 0
]
```
